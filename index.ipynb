```python
import pulp as pp
```

## 1. **Initialize the MODEL**:

#Define the name & Problem Type.

```python
model = pp.LpProblem(name='vitamin-problem', # name
                     sense=pp.LpMinimize) # problem type
```

## 2. **Declare the VARIABLES**:


The supplement model consists of the following decision variables:

```python
# how much SuperVit?
supervit = pp.LpVariable(name="supervit",  # define the name
                    lowBound=0,  # ensure non-negativity
                    cat='Integer') #Integer value since we are working with quantity.

# how much NewHealth?
newhealth = pp.LpVariable(name="newhealth", # define the name
                 lowBound=0, # ensure non-negativity
                 cat='Integer') #Integer value since we are working with quantity.
```

## 3. **Create function to OPTIMIZE**:

This function consists of the linear combination of the variables and their _given coefficients_:

```python
supervitCoeff=0.2 #supervit coefficient
newhealthCoeff=0.3 #newhealth coefficient
obj_func = supervitCoeff*supervit + newhealthCoeff*newhealth #create the function that will calculate the problem's objective solution.
```

## 4. **Represent the constraints**:

These are the rules the model (set of variables) must obey:

```python
C1= pp.LpConstraint(name='Vitamin C',   # define the name
                    e= 20 * supervit + 30 * newhealth, #rule
                    rhs=60, #Constraint
                    sense=pp.LpConstraintGE) # >= 'sense'
                    #Same logic for the entire constraints listed below.
C2= pp.LpConstraint(name='Calcium',
                    e= 500 * supervit + 250 * newhealth,
                    rhs=1000,
                    sense=pp.LpConstraintGE)
C3= pp.LpConstraint(name='Iron',
                    e= 9 * supervit + 2 * newhealth,
                    rhs=18,
                    sense=pp.LpConstraintGE)
C4= pp.LpConstraint(name='Niacin',
                    e= 2 * supervit + 10 * newhealth,
                    rhs=20,
                    sense=pp.LpConstraintGE)
C5= pp.LpConstraint(name='Magnesium',
                    e= 60 * supervit + 90 * newhealth,
                    rhs=360,
                    sense=pp.LpConstraintGE)
```

## 5. **Build MODEL**:

add up the objective functions and all constraints.

```python
model = pp.LpProblem(name='vitamin-problem', sense=pp.LpMinimize)

model += obj_func
model += C1
model += C2
model += C3
model += C4
model += C5
```

## 6. **Solve the MODEL**:



```python
# model.solve()
model.solve(pp.PULP_CBC_CMD(msg=0)) #Execute to solve the model problem.
```
